## 4. 移動成功時だけ状態を更新する / Update state only on successful move

短い説明: 成功時だけ `x/y/turn` を更新します。
Short note: update `x/y/turn` only when movement succeeds.

```python
def move_player(casino_map, player, movement):
    dx, dy = movement
    next_x = player['x'] + dx
    next_y = player['y'] + dy
    if (dx == 0 and dy == 0) or not can_move(casino_map, next_x, next_y):
        return False
    player['x'] = next_x
    player['y'] = next_y
    player['turn'] += 1
    return True
```

### 実行前の予想欄 / Prediction before run
- 壁にぶつかったとき `turn` は?
- 右へ進めたとき `turn` は?

```python
p = {'x': 1, 'y': 1, 'turn': 0}
assert move_player(casino_map, p, (0, -1)) is False
assert p == {'x': 1, 'y': 1, 'turn': 0}
assert move_player(casino_map, p, (1, 0)) is True
assert p == {'x': 2, 'y': 1, 'turn': 1}
```
