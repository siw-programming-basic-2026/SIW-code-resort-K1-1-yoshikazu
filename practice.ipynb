## 1. 辞書の作成・参照・更新 / Create, read and update a dictionary

### キーと値を扱う / Work with keys and values

辞書は「キー」と「値」の組でデータを管理します。文字列のキーには引用符が必要です。
A dictionary stores key/value pairs. String keys need quotation marks.

````python
player = {"name": "Alex", "money": 1000, "x": 1, "y": 1}
print(player["name"])
print(player["money"])

player["money"] -= 20
player["focus"] = 50
print(player["money"])
print(player.get("stamina", 0))
````

### 実行結果と更新・追加 / Output, updates and additions

出力は順に `Alex`、`1000`、`980`、`0` です。既存のキーへの代入は更新、存在しないキーへの代入は追加になります。
The output is `Alex`, `1000`, `980` and `0`. Assigning to an existing key updates it; assigning to a new key adds it.

### 存在しないキーとget / Missing keys and get

`player["stamina"]` のように存在しないキーを参照すると `KeyError` になります。`get(key, default)` は、キーがなければ指定した既定値を返します。辞書への追加は行いません。
Accessing a missing key with `player["stamina"]` raises `KeyError`. `get(key, default)` returns the default when the key is missing, without adding it to the dictionary.

### 新しいキーを追加する / Add a new key

新しい項目を追加するには、`辞書[キー] = 値` と代入します。存在しないキーへの読み取りはエラーになりますが、代入は新しいキーと値の追加になります。
To add an entry, assign a value with `dictionary[key] = value`. Reading a missing key raises an error, but assigning to it adds the new key/value pair.

````python
print("stamina" in player)
print(player.get("stamina", 0))
print("stamina" in player)

player["stamina"] = 50
print("stamina" in player)
print(player["stamina"])
````

出力は順に `False`、`0`、`False`、`True`、`50` です。`in` はキーが存在するかを調べます。`get()` の後はまだ存在しませんが、代入すると追加されます。同じキーにもう一度代入すると、既存の値を更新します。
The output is `False`, `0`, `False`, `True` and `50`. `in` checks whether a key exists. The key is still absent after `get()`, but assignment adds it. Assigning to the same key again updates its existing value.

### itemsで繰り返す / Iterate with items

キーと値を順番に取り出すには `items()` を使います。
Use `items()` to iterate over keys and values together.

````python
for key, value in player.items():
    print(key, value)
````

## 2. 処理を関数にする / Put operations into functions

### 位置引数とキーワード引数 / Positional and keyword arguments

まずは、必要な引数を名前で明示する関数から始めます。呼び出し時の位置引数は順番で、キーワード引数は名前で対応します。
Start with functions that explicitly name the arguments they need. Positional arguments match by order; keyword arguments match by name.

````python
def create_player(name, money=1000):
    return {"name": name, "money": money, "x": 1, "y": 1}


player = create_player("Alex")
guest = create_player(money=500, name="Sam")

````

````python
print(player,guest)
````

### 既定値と戻り値 / Defaults and return values

`money=1000` は引数を省略したときの既定値です。`return` は結果を呼び出し元へ返します。
`money=1000` supplies a default when the argument is omitted. `return` sends the result back to the caller.

### 辞書を丸ごと渡す / Pass the whole dictionary

辞書全体を1つの引数として渡せば、関数内で複数の項目を扱えます。
Pass the entire dictionary as one argument when the function needs several fields.

````python
def pay(player, price):
    if player["money"] < price:
        return False
    player["money"] -= price
    return True


print(pay(guest, 50))
print(guest["money"])
````

### 辞書の更新が伝わる仕組み / How dictionary updates reach the caller

出力は `True`、`980` です。引数の `player` は呼び出し元と同じ辞書を参照するため、項目の更新が呼び出し元にも反映されます。関数内で `player = {}` と代入し直すだけなら、呼び出し元の変数は置き換わりません。
The output is `True` and `980`. The parameter references the same dictionary as the caller, so updating its entries is visible to the caller. Reassigning the local name with `player = {}` does not replace the caller's variable.

## 3. 定義で使う `*args` / Collect positional arguments

### 位置引数をまとめる仕組み / How positional arguments are collected

関数定義の `*args` は、任意の個数の位置引数をタプルにまとめます。`args` は慣例的な名前で、働きを決めるのは `*` です。
In a function definition, `*args` collects any number of positional arguments into a tuple. `args` is a conventional name; the `*` gives it this behavior.

````python
def total_cost(*prices):
    return sum(prices)


print(total_cost(20, 30))
print(total_cost(20, 30, 100))
print(total_cost())
````

### タプルの中身と実行結果 / Tuple contents and output

出力は `50`、`150`、`0` です。最初の呼び出しでは `prices` は `(20, 30)` になります。
The output is `50`, `150` and `0`. In the first call, `prices` is `(20, 30)`.

## 4. 定義で使う `**kwargs` / Collect keyword arguments

### キーワード引数をまとめる仕組み / How keyword arguments are collected

関数定義の `**kwargs` は、明示された引数に対応しないキーワード引数を辞書にまとめます。`kwargs` も慣例的な名前です。
In a function definition, `**kwargs` collects keyword arguments that do not match explicitly declared parameters into a dictionary. `kwargs` is also a conventional name.

````python
def describe_player(name, **details):
    print(name)
    print(details)


describe_player("Alex", money=1000, focus=50)
````

````python
describe_player("Yoshikazu",weight=100,age=50)
````

### 辞書に入る項目 / Entries in the collected dictionary

`name` は `"Alex"`、`details` は `{"money": 1000, "focus": 50}` になります。
`name` is `"Alex"`; `details` is `{"money": 1000, "focus": 50}`.

### 明示的な引数を使う場面 / When to use explicit parameters

任意のキーを受け付けるため、キーの綴り間違いも受け取ってしまいます。必須項目が決まっている関数では、通常の名前付きの引数を優先しましょう。
Because arbitrary keys are accepted, misspelled keys are accepted too. Prefer explicit parameters when the required fields are known.

## 5. 呼び出しで使う `*` と `**` / Unpack arguments in a call

### 定義での収集と呼び出しでの展開 / Collection versus unpacking

関数の呼び出しでは逆に、`*` はリストやタプルなどの要素を位置引数へ、`**` は辞書などのマッピングの項目をキーワード引数へ展開します。呼び出す関数の定義に `*args` や `**kwargs` がなくても使えます。
In a function call, `*` unpacks an iterable such as a list or tuple into positional arguments, while `**` unpacks a mapping such as a dictionary into keyword arguments. The called function does not need to declare `*args` or `**kwargs`.

````python
def next_position(x, y, dx, dy):
    return x + dx, y + dy


movement = (1, 0)
print(next_position(1, 1, *movement))

settings = {"name": "Sam", "money": 500}
guest = create_player(**settings)
print(guest["name"], guest["money"])

print(guest)
````

### 展開後の引数と実行結果 / Expanded arguments and output

出力は `(2, 1)` と `Sam 500` です。`*movement` は `1, 0` を渡すのと同じで、`**settings` は `name="Sam", money=500` を渡すのと同じです。
The output is `(2, 1)` and `Sam 500`. `*movement` is equivalent to passing `1, 0`; `**settings` is equivalent to passing `name="Sam", money=500`.

### 展開時の注意点 / Unpacking constraints

`**` で展開するキーは文字列である必要があります。関数が受け付けない名前や、同じ引数への二重指定は `TypeError` になります。たとえば `create_player(**player)` は `x` と `y` を受け付けないため失敗します。辞書全体を渡す `pay(player, 20)` と、項目を引数へ展開する呼び出しは区別しましょう。
Keys unpacked with `**` must be strings. Unsupported names or multiple values for the same argument raise `TypeError`. For example, `create_player(**player)` fails because the function does not accept x or y. Distinguish passing the dictionary itself, as in `pay(player, 20)`, from unpacking its entries into arguments.

## 6. 引数を渡す方法の整理 / Argument reference

### 書き方の比較 / Syntax comparison

| 書き方 / Syntax | 意味 / Meaning |
| --- | --- |
| `def f(player):` | 辞書などを1つの値として受け取る / Receive one value, such as a dictionary |
| `def f(*args):` | 位置引数をタプルにまとめる / Collect positional arguments into a tuple |
| `def f(**kwargs):` | キーワード引数を辞書にまとめる / Collect keyword arguments into a dictionary |
| `f(*values)` | 要素を位置引数に展開する / Unpack elements into positional arguments |
| `f(**settings)` | 項目をキーワード引数に展開する / Unpack entries into keyword arguments |

### 補足：キーワード専用引数 / Additional syntax: keyword-only arguments

補足：定義で単独の `*` を使うと、その後の引数はキーワードでのみ指定できます。
Additional syntax: a bare `*` in a definition makes subsequent parameters keyword-only.

````python
def recover(player, *, amount=20):
    player["focus"] = min(100, player["focus"] + amount)


player["focus"] = 50
recover(player, amount=10)
print(player["focus"])
````

出力は `60` です。`recover(player, 10)` は `TypeError` になります。引数の役割を明示したいときに使います。
The output is `60`. `recover(player, 10)` raises `TypeError`. Use this syntax when you want callers to name an argument explicitly.
